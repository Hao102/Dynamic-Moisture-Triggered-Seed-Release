```python
# ============================================================
# MODEL 1 — REAL-TIME SENSOR
# MULTIPLE LINEAR REGRESSION
# ============================================================
#
# Method: Periodic Batch Retraining
#
# LinearRegression does not support partial_fit(), so the model
# is periodically retrained when new sensor data become available.
#
# Workflow:
# Initial 2/3 training data
#       ↓
# Predict new sensor batch
#       ↓
# Receive actual sensor values
#       ↓
# Add new data to training set
#       ↓
# Retrain model
#       ↓
# Predict next sensor batch
#
# IMPORTANT:
# Prediction is made BEFORE the new observations are used
# for model updating to avoid data leakage.
# ============================================================


# ------------------------------------------------------------
# 1. Import libraries
# ------------------------------------------------------------

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time

from sklearn.linear_model import LinearRegression

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ------------------------------------------------------------
# 2. Real-time update settings
# ------------------------------------------------------------
#
# X_train / y_train = initial 2/3 training data
# X_test / y_test   = simulated incoming sensor data
#
# The model is updated every 100 new observations.
# ------------------------------------------------------------

UPDATE_BATCH_SIZE = 100

print("=" * 65)
print("MODEL 1 — REAL-TIME SENSOR")
print("=" * 65)

print("\nMethod: Periodic Batch Retraining")

print(
    "Initial training observations:",
    len(X_train)
)

print(
    "Sensor observations:",
    len(X_test)
)

print(
    "Update batch size:",
    UPDATE_BATCH_SIZE
)


# ------------------------------------------------------------
# 3. Static model baseline
# ------------------------------------------------------------
#
# The original Linear Regression from Part 2 is used as
# the static baseline. It does not learn from new sensor data.
# ------------------------------------------------------------

static_mae = linear_mae
static_rmse = linear_rmse
static_r2 = linear_r2


print("\n" + "=" * 65)
print("STATIC LINEAR MODEL")
print("=" * 65)

print("MAE  :", round(static_mae, 4))
print("RMSE :", round(static_rmse, 4))
print("R²   :", round(static_r2, 4))


# ------------------------------------------------------------
# 4. Create adaptive model
# ------------------------------------------------------------
#
# Start with the same 2/3 training data as the static model.
# ------------------------------------------------------------

adaptive_linear_model = LinearRegression()

adaptive_linear_model.fit(
    X_train,
    y_train
)


# Training data will grow as new sensor observations arrive.

X_adaptive_train = X_train.copy()
y_adaptive_train = y_train.copy()


# ------------------------------------------------------------
# 5. Storage
# ------------------------------------------------------------

adaptive_predictions = []
adaptive_actual = []

update_points = []
update_times = []


# ------------------------------------------------------------
# 6. Real-time sensor simulation
# ------------------------------------------------------------
#
# Each batch represents newly arriving sensor information.
#
# For each batch:
# 1. Predict using the current model
# 2. Receive the actual sensor target
# 3. Add new observations to training data
# 4. Retrain the model for the next batch
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("STARTING SENSOR SIMULATION")
print("=" * 65)


for start in range(
    0,
    len(X_test),
    UPDATE_BATCH_SIZE
):

    end = min(
        start + UPDATE_BATCH_SIZE,
        len(X_test)
    )


    # New sensor batch

    X_sensor_batch = X_test.iloc[
        start:end
    ]

    y_sensor_batch = y_test.iloc[
        start:end
    ]


    # --------------------------------------------------------
    # STEP 1 — Predict BEFORE updating
    # --------------------------------------------------------

    batch_predictions = adaptive_linear_model.predict(
        X_sensor_batch
    )


    adaptive_predictions.extend(
        batch_predictions
    )

    adaptive_actual.extend(
        y_sensor_batch.values
    )


    # --------------------------------------------------------
    # STEP 2 — Add new sensor information
    # --------------------------------------------------------
    #
    # After prediction, the actual sensor values are available
    # and can be used to improve future predictions.
    # --------------------------------------------------------

    X_adaptive_train = pd.concat(
        [
            X_adaptive_train,
            X_sensor_batch
        ],
        ignore_index=True
    )


    y_adaptive_train = pd.concat(
        [
            y_adaptive_train.reset_index(drop=True),
            y_sensor_batch.reset_index(drop=True)
        ],
        ignore_index=True
    )


    # --------------------------------------------------------
    # STEP 3 — Retrain model
    # --------------------------------------------------------
    #
    # LinearRegression cannot update incrementally, so it is
    # fitted again using historical + new sensor observations.
    # --------------------------------------------------------

    update_start = time.time()


    adaptive_linear_model.fit(
        X_adaptive_train,
        y_adaptive_train
    )


    update_time = time.time() - update_start


    update_times.append(
        update_time
    )

    update_points.append(
        end
    )


# ------------------------------------------------------------
# 7. Evaluate adaptive model
# ------------------------------------------------------------

adaptive_predictions = np.array(
    adaptive_predictions
)

adaptive_actual = np.array(
    adaptive_actual
)


adaptive_mae = mean_absolute_error(
    adaptive_actual,
    adaptive_predictions
)


adaptive_rmse = np.sqrt(
    mean_squared_error(
        adaptive_actual,
        adaptive_predictions
    )
)


adaptive_r2 = r2_score(
    adaptive_actual,
    adaptive_predictions
)


print("\n" + "=" * 65)
print("REAL-TIME UPDATED LINEAR MODEL")
print("=" * 65)

print("MAE  :", round(adaptive_mae, 4))
print("RMSE :", round(adaptive_rmse, 4))
print("R²   :", round(adaptive_r2, 4))


# ------------------------------------------------------------
# 8. Static vs Adaptive comparison
# ------------------------------------------------------------

linear_realtime_comparison = pd.DataFrame({

    "Method": [
        "Static Linear Regression",
        "Real-Time Updated Linear Regression"
    ],

    "MAE": [
        static_mae,
        adaptive_mae
    ],

    "RMSE": [
        static_rmse,
        adaptive_rmse
    ],

    "R²": [
        static_r2,
        adaptive_r2
    ]

})


print("\n" + "=" * 65)
print("STATIC VS REAL-TIME UPDATED MODEL")
print("=" * 65)

display(
    linear_realtime_comparison.round(4)
)


# ------------------------------------------------------------
# 9. Performance change
# ------------------------------------------------------------

mae_improvement = (
    static_mae - adaptive_mae
)

rmse_improvement = (
    static_rmse - adaptive_rmse
)

r2_change = (
    adaptive_r2 - static_r2
)


print("\nPerformance Change:")

print(
    "MAE improvement :",
    round(mae_improvement, 4)
)

print(
    "RMSE improvement:",
    round(rmse_improvement, 4)
)

print(
    "R² change       :",
    round(r2_change, 4)
)


# ------------------------------------------------------------
# 10. Model update time
# ------------------------------------------------------------
#
# Update time is important for assessing whether the method
# can operate fast enough for a real-time seeding system.
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("UPDATE PERFORMANCE")
print("=" * 65)

print(
    "Number of updates:",
    len(update_times)
)

print(
    "Average update time:",
    round(
        np.mean(update_times),
        6
    ),
    "seconds"
)

print(
    "Maximum update time:",
    round(
        np.max(update_times),
        6
    ),
    "seconds"
)


# ------------------------------------------------------------
# 11. Static vs Adaptive predictions
# ------------------------------------------------------------

plot_size = min(
    500,
    len(y_test)
)


plt.figure(figsize=(11, 5))


plt.plot(
    y_test.values[:plot_size],
    label="Actual Sensor Moisture"
)


plt.plot(
    linear_pred[:plot_size],
    label="Static Model"
)


plt.plot(
    adaptive_predictions[:plot_size],
    label="Real-Time Updated Model"
)


plt.xlabel(
    "Sensor Observation"
)

plt.ylabel(
    "Moisture (VWC %)"
)

plt.title(
    "Linear Regression — Static vs Real-Time Updating"
)

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 12. Real-time prediction error
# ------------------------------------------------------------
#
# Vertical lines indicate where model updates occurred.
# The moving average shows the overall error trend.
# ------------------------------------------------------------

adaptive_error = np.abs(
    adaptive_actual
    -
    adaptive_predictions
)


moving_error = (
    pd.Series(adaptive_error)
    .rolling(
        window=100,
        min_periods=1
    )
    .mean()
)


plt.figure(figsize=(11, 5))


plt.plot(
    adaptive_error,
    alpha=0.4,
    label="Absolute Error"
)


plt.plot(
    moving_error,
    linewidth=2,
    label="100-Observation Moving Average"
)


for update_point in update_points:

    plt.axvline(
        x=update_point,
        linestyle="--",
        alpha=0.15
    )


plt.xlabel(
    "Sensor Observation"
)

plt.ylabel(
    "Absolute Error (VWC %)"
)

plt.title(
    "Linear Regression — Real-Time Prediction Error"
)

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 13. Final summary
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("REAL-TIME LINEAR REGRESSION SUMMARY")
print("=" * 65)

print(
    "Update method:",
    "Periodic Batch Retraining"
)

print(
    "Update every:",
    UPDATE_BATCH_SIZE,
    "sensor observations"
)

print(
    "Total updates:",
    len(update_points)
)

print(
    "\nStatic RMSE:",
    round(static_rmse, 4)
)

print(
    "Adaptive RMSE:",
    round(adaptive_rmse, 4)
)

print(
    "\nStatic R²:",
    round(static_r2, 4)
)

print(
    "Adaptive R²:",
    round(adaptive_r2, 4)
)

print(
    "\nModel 1 real-time sensor completed."
)
```

Method = Periodic Batch Retraining → update every 100 sensor observations → predict first → receive sensor feedback → retrain → use updated model for next batch.

### Method — Periodic Batch Retraining

Multiple Linear Regression does not support direct incremental updating through `partial_fit()`. Therefore, **Periodic Batch Retraining** is used to incorporate new sensor information into the model.

The model is initially trained using **2/3 of the dataset**, while the remaining **1/3 is used to simulate incoming sensor observations**. Sensor data are processed in batches of **100 observations**.

For each batch, the current Linear Regression model first predicts soil moisture. Once the actual sensor values become available, the new observations are added to the existing training dataset and the model is retrained. The updated model is then used to predict the next sensor batch.

The real-time process is:

**Initial Training → Sensor Batch → Prediction → Sensor Feedback → Add New Data → Retrain Model → Next Sensor Batch**

This ensures that new sensor information is used only **after prediction**, avoiding data leakage. The static and real-time models are compared using **MAE, RMSE, R², and model update time**.

## Model 2 — Random Forest Regression with Real-Time Sensor:

```python
# ============================================================
# MODEL 2 — RANDOM FOREST REAL-TIME
# ============================================================
#
# Method: Periodic Batch Retraining
#
# RandomForestRegressor does not support partial_fit().
# Therefore, the model is retrained periodically using new
# sensor observations.
#
# Workflow:
# Initial 2/3 training data
#       ↓
# Predict new sensor batch
#       ↓
# Receive actual sensor values
#       ↓
# Add new observations to training data
#       ↓
# Retrain Random Forest
#       ↓
# Predict next sensor batch
#
# Prediction occurs BEFORE retraining to avoid data leakage.
# ============================================================


# ------------------------------------------------------------
# 1. Import libraries
# ------------------------------------------------------------

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time

from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ------------------------------------------------------------
# 2. Real-time settings
# ------------------------------------------------------------
#
# X_train / y_train = initial 2/3 training data
# X_test / y_test   = simulated incoming sensor observations
#
# No new train/test split is performed.
# ------------------------------------------------------------

RF_UPDATE_BATCH_SIZE = 100


print("=" * 65)
print("MODEL 2 — RANDOM FOREST REAL-TIME")
print("=" * 65)

print("\nMethod: Periodic Batch Retraining")

print(
    "Initial training observations:",
    len(X_train)
)

print(
    "Sensor observations:",
    len(X_test)
)

print(
    "Update batch size:",
    RF_UPDATE_BATCH_SIZE
)


# ------------------------------------------------------------
# 3. Create static Random Forest
# ------------------------------------------------------------
#
# The static model is trained once on the original 2/3
# training data and is NOT updated with sensor information.
#
# This provides the baseline for comparison.
# ------------------------------------------------------------

rf_static_model = RandomForestRegressor(
    n_estimators=200,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)


rf_static_model.fit(
    X_train,
    y_train
)


# Static prediction on the full sensor/test set

rf_static_pred = rf_static_model.predict(
    X_test
)


# ------------------------------------------------------------
# 4. Calculate static performance
# ------------------------------------------------------------

rf_static_mae = mean_absolute_error(
    y_test,
    rf_static_pred
)


rf_static_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        rf_static_pred
    )
)


rf_static_r2 = r2_score(
    y_test,
    rf_static_pred
)


print("\n" + "=" * 65)
print("STATIC RANDOM FOREST")
print("=" * 65)

print(
    "MAE  :",
    round(rf_static_mae, 4)
)

print(
    "RMSE :",
    round(rf_static_rmse, 4)
)

print(
    "R²   :",
    round(rf_static_r2, 4)
)


# ------------------------------------------------------------
# 5. Create adaptive Random Forest
# ------------------------------------------------------------
#
# The adaptive model starts with the same initial training data.
#
# Unlike the static model, it will learn from new sensor
# observations after each batch.
# ------------------------------------------------------------

adaptive_rf_model = RandomForestRegressor(
    n_estimators=200,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)


adaptive_rf_model.fit(
    X_train,
    y_train
)


# Expandable training dataset

X_rf_adaptive_train = X_train.copy()

y_rf_adaptive_train = y_train.copy()


# ------------------------------------------------------------
# 6. Storage
# ------------------------------------------------------------

rf_adaptive_predictions = []

rf_adaptive_actual = []

rf_update_points = []

rf_update_times = []


# ------------------------------------------------------------
# 7. Real-time sensor simulation
# ------------------------------------------------------------
#
# For each batch:
#
# 1. Predict using current Random Forest
# 2. Receive actual sensor values
# 3. Add sensor observations to training data
# 4. Retrain Random Forest
# 5. Use updated model for the next batch
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("STARTING SENSOR SIMULATION")
print("=" * 65)


for start in range(
    0,
    len(X_test),
    RF_UPDATE_BATCH_SIZE
):

    end = min(
        start + RF_UPDATE_BATCH_SIZE,
        len(X_test)
    )


    # New sensor batch

    X_sensor_batch = X_test.iloc[
        start:end
    ]

    y_sensor_batch = y_test.iloc[
        start:end
    ]


    # --------------------------------------------------------
    # STEP 1 — Predict BEFORE retraining
    # --------------------------------------------------------

    batch_predictions = adaptive_rf_model.predict(
        X_sensor_batch
    )


    rf_adaptive_predictions.extend(
        batch_predictions
    )


    rf_adaptive_actual.extend(
        y_sensor_batch.values
    )


    # --------------------------------------------------------
    # STEP 2 — Add new sensor information
    # --------------------------------------------------------
    #
    # After prediction, actual sensor values become available
    # and can be used to improve future predictions.
    # --------------------------------------------------------

    X_rf_adaptive_train = pd.concat(
        [
            X_rf_adaptive_train,
            X_sensor_batch
        ],
        ignore_index=True
    )


    y_rf_adaptive_train = pd.concat(
        [
            y_rf_adaptive_train.reset_index(
                drop=True
            ),

            y_sensor_batch.reset_index(
                drop=True
            )
        ],
        ignore_index=True
    )


    # --------------------------------------------------------
    # STEP 3 — Periodic Batch Retraining
    # --------------------------------------------------------
    #
    # Random Forest cannot use partial_fit(), so the complete
    # model is retrained using historical + new sensor data.
    # --------------------------------------------------------

    update_start = time.time()


    adaptive_rf_model.fit(
        X_rf_adaptive_train,
        y_rf_adaptive_train
    )


    update_time = (
        time.time()
        -
        update_start
    )


    rf_update_times.append(
        update_time
    )


    rf_update_points.append(
        end
    )


# ------------------------------------------------------------
# 8. Convert results to arrays
# ------------------------------------------------------------

rf_adaptive_predictions = np.array(
    rf_adaptive_predictions
)


rf_adaptive_actual = np.array(
    rf_adaptive_actual
)


# ------------------------------------------------------------
# 9. Evaluate adaptive Random Forest
# ------------------------------------------------------------

rf_adaptive_mae = mean_absolute_error(
    rf_adaptive_actual,
    rf_adaptive_predictions
)


rf_adaptive_rmse = np.sqrt(
    mean_squared_error(
        rf_adaptive_actual,
        rf_adaptive_predictions
    )
)


rf_adaptive_r2 = r2_score(
    rf_adaptive_actual,
    rf_adaptive_predictions
)


print("\n" + "=" * 65)
print("REAL-TIME RANDOM FOREST")
print("=" * 65)

print(
    "MAE  :",
    round(rf_adaptive_mae, 4)
)

print(
    "RMSE :",
    round(rf_adaptive_rmse, 4)
)

print(
    "R²   :",
    round(rf_adaptive_r2, 4)
)


# ------------------------------------------------------------
# 10. Static vs Real-Time comparison
# ------------------------------------------------------------

rf_realtime_comparison = pd.DataFrame({

    "Method": [
        "Static Random Forest",
        "Real-Time Random Forest"
    ],

    "MAE": [
        rf_static_mae,
        rf_adaptive_mae
    ],

    "RMSE": [
        rf_static_rmse,
        rf_adaptive_rmse
    ],

    "R²": [
        rf_static_r2,
        rf_adaptive_r2
    ]

})


print("\n" + "=" * 65)
print("STATIC VS REAL-TIME RANDOM FOREST")
print("=" * 65)


display(
    rf_realtime_comparison.round(4)
)


# ------------------------------------------------------------
# 11. Performance change
# ------------------------------------------------------------

rf_mae_improvement = (
    rf_static_mae
    -
    rf_adaptive_mae
)


rf_rmse_improvement = (
    rf_static_rmse
    -
    rf_adaptive_rmse
)


rf_r2_change = (
    rf_adaptive_r2
    -
    rf_static_r2
)


print("\nPerformance Change:")


print(
    "MAE improvement :",
    round(rf_mae_improvement, 4)
)


print(
    "RMSE improvement:",
    round(rf_rmse_improvement, 4)
)


print(
    "R² change       :",
    round(rf_r2_change, 4)
)


# ------------------------------------------------------------
# 12. Update performance
# ------------------------------------------------------------
#
# Random Forest is more computationally expensive than
# Linear Regression, so update time is also evaluated.
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("UPDATE PERFORMANCE")
print("=" * 65)


print(
    "Number of updates:",
    len(rf_update_times)
)


print(
    "Average update time:",
    round(
        np.mean(rf_update_times),
        6
    ),
    "seconds"
)


print(
    "Maximum update time:",
    round(
        np.max(rf_update_times),
        6
    ),
    "seconds"
)


# ------------------------------------------------------------
# 13. Static vs Real-Time predictions
# ------------------------------------------------------------

plot_size = min(
    500,
    len(y_test)
)


plt.figure(figsize=(11, 5))


plt.plot(
    y_test.values[:plot_size],
    label="Actual Sensor Moisture"
)


plt.plot(
    rf_static_pred[:plot_size],
    label="Static Random Forest"
)


plt.plot(
    rf_adaptive_predictions[:plot_size],
    label="Real-Time Random Forest"
)


plt.xlabel(
    "Sensor Observation"
)

plt.ylabel(
    "Moisture (VWC %)"
)

plt.title(
    "Random Forest — Static vs Real-Time"
)


plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 14. Real-time prediction error
# ------------------------------------------------------------

rf_adaptive_error = np.abs(
    rf_adaptive_actual
    -
    rf_adaptive_predictions
)


rf_moving_error = (
    pd.Series(rf_adaptive_error)
    .rolling(
        window=100,
        min_periods=1
    )
    .mean()
)


plt.figure(figsize=(11, 5))


plt.plot(
    rf_adaptive_error,
    alpha=0.4,
    label="Absolute Error"
)


plt.plot(
    rf_moving_error,
    linewidth=2,
    label="100-Observation Moving Average"
)


# Vertical lines show model retraining points

for update_point in rf_update_points:

    plt.axvline(
        x=update_point,
        linestyle="--",
        alpha=0.15
    )


plt.xlabel(
    "Sensor Observation"
)

plt.ylabel(
    "Absolute Error (VWC %)"
)

plt.title(
    "Random Forest — Real-Time Prediction Error"
)


plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 15. Final summary
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("RANDOM FOREST REAL-TIME SUMMARY")
print("=" * 65)


print(
    "Method:",
    "Periodic Batch Retraining"
)


print(
    "Update every:",
    RF_UPDATE_BATCH_SIZE,
    "sensor observations"
)


print(
    "Total updates:",
    len(rf_update_points)
)


print(
    "\nStatic RMSE:",
    round(rf_static_rmse, 4)
)


print(
    "Real-Time RMSE:",
    round(rf_adaptive_rmse, 4)
)


print(
    "\nStatic R²:",
    round(rf_static_r2, 4)
)


print(
    "Real-Time R²:",
    round(rf_adaptive_r2, 4)
)


print(
    "\nModel 2 — Random Forest Real-Time completed."
)
```

### Method — Periodic Batch Retraining

Random Forest does not support direct incremental learning through `partial_fit()`. Therefore, **Periodic Batch Retraining** is used to incorporate new sensor information into the model.

The model is initially trained using **2/3 of the dataset**. The remaining **1/3 is used to simulate incoming sensor observations**. Sensor data are processed in batches of **100 observations**.

For each batch, the current Random Forest model first predicts soil moisture before seeing the actual target values. Once the actual sensor observations become available, they are added to the existing training dataset and the Random Forest model is retrained. The updated model is then used to predict the next sensor batch.

The real-time process is:

**Initial Training → Sensor Batch → Prediction → Sensor Feedback → Add New Data → Retrain Model → Next Sensor Batch**

This ensures that new sensor information is only used **after prediction**, reducing the risk of data leakage. Model performance is evaluated by comparing the static and real-time models using **MAE, RMSE, R², and model update time**.

### Method — Mini-Batch Fine-Tuning

Unlike Linear Regression and Random Forest, the trained CNN can continue learning from new data by updating its existing network weights. Therefore, **Mini-Batch Fine-Tuning** is used to incorporate new sensor information.

The CNN is initially trained using **2/3 of the dataset**, while the remaining **1/3 is used to simulate incoming sensor observations**. Sensor data are processed in batches of **100 observations**.

For each batch, the current CNN first predicts soil moisture before seeing the actual target values. Once the actual sensor values become available, the same batch is used to fine-tune the existing CNN for a small number of additional epochs. The updated CNN is then used to predict the next sensor batch.

The real-time process is:

**Initial CNN → Sensor Batch → Prediction → Sensor Feedback → Mini-Batch Fine-Tuning → Next Sensor Batch**

Unlike periodic full retraining, the CNN retains its previously learned weights and updates them using the latest sensor information. The static and real-time CNN models are compared using **MAE, RMSE, R², and model update time**.

## Model 4 — Mixture Gaussian Process Regression Real-Time:

```python
# ============================================================
# MODEL 4 — MIXTURE GPR REAL-TIME
# ============================================================
#
# Method:
# Initial Kernel Optimisation + Sliding-Window Fast Refitting
#
# Initial GPR optimises the kernel once.
# New sensor data arrive in batches of 100.
# GPR is then refitted without kernel re-optimisation.
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time
import warnings

from sklearn.preprocessing import StandardScaler
from sklearn.mixture import GaussianMixture
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import ConstantKernel, RBF, WhiteKernel
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.exceptions import ConvergenceWarning


# ------------------------------------------------------------
# 1. Settings
# ------------------------------------------------------------

UPDATE_SIZE = 100
WINDOW_SIZE = 1000
N_REGIMES = 3

print("=" * 65)
print("MODEL 4 — MIXTURE GPR REAL-TIME")
print("=" * 65)
print("Method: Initial Kernel Optimisation + Sliding-Window Fast Refitting")
print("Update every:", UPDATE_SIZE, "observations")
print("Window size:", WINDOW_SIZE, "per regime")


# ------------------------------------------------------------
# 2. Standardise data
# ------------------------------------------------------------

scaler_gpr = StandardScaler()

X_train_gpr = scaler_gpr.fit_transform(X_train)
X_test_gpr = scaler_gpr.transform(X_test)


# ------------------------------------------------------------
# 3. Identify Gaussian regimes
# ------------------------------------------------------------

gmm = GaussianMixture(
    n_components=N_REGIMES,
    random_state=42
)

gmm.fit(X_train_gpr)

train_regimes = gmm.predict(X_train_gpr)
test_regimes = gmm.predict(X_test_gpr)


# ------------------------------------------------------------
# 4. GPR functions
# ------------------------------------------------------------

def initial_gpr():

    kernel = (
        ConstantKernel(1.0)
        * RBF(np.ones(len(features)))
        + WhiteKernel(0.001)
    )

    return GaussianProcessRegressor(
        kernel=kernel,
        normalize_y=True,
        random_state=42
    )


def realtime_gpr(kernel):

    return GaussianProcessRegressor(
        kernel=kernel,
        normalize_y=True,
        optimizer=None,
        random_state=42
    )


# ------------------------------------------------------------
# 5. Train initial GPR for each regime
# ------------------------------------------------------------

rng = np.random.RandomState(42)

regime_X = {}
regime_y = {}
models = {}
kernels = {}


for r in range(N_REGIMES):

    idx = np.where(train_regimes == r)[0]

    if len(idx) > WINDOW_SIZE:
        idx = rng.choice(
            idx,
            WINDOW_SIZE,
            replace=False
        )

    regime_X[r] = X_train_gpr[idx].copy()
    regime_y[r] = y_train.iloc[idx].values.copy()

    model = initial_gpr()

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", ConvergenceWarning)

        model.fit(
            regime_X[r],
            regime_y[r]
        )

    models[r] = model
    kernels[r] = model.kernel_


print("\nInitial Mixture GPR trained.")


# ------------------------------------------------------------
# 6. Static GPR baseline
# ------------------------------------------------------------

static_pred = np.zeros(len(X_test))
static_std = np.zeros(len(X_test))


for r in range(N_REGIMES):

    idx = np.where(test_regimes == r)[0]

    if len(idx) == 0:
        continue

    pred, std = models[r].predict(
        X_test_gpr[idx],
        return_std=True
    )

    static_pred[idx] = pred
    static_std[idx] = std


static_mae = mean_absolute_error(
    y_test,
    static_pred
)

static_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        static_pred
    )
)

static_r2 = r2_score(
    y_test,
    static_pred
)


# ------------------------------------------------------------
# 7. Create adaptive models
# ------------------------------------------------------------

adaptive_X = {
    r: regime_X[r].copy()
    for r in range(N_REGIMES)
}

adaptive_y = {
    r: regime_y[r].copy()
    for r in range(N_REGIMES)
}

adaptive_models = {}


for r in range(N_REGIMES):

    adaptive_models[r] = realtime_gpr(
        kernels[r]
    )

    adaptive_models[r].fit(
        adaptive_X[r],
        adaptive_y[r]
    )


# ------------------------------------------------------------
# 8. Real-time sensor simulation
# ------------------------------------------------------------
#
# Predict → sensor feedback → sliding window → refit
# ------------------------------------------------------------

rt_pred = []
rt_actual = []
rt_std = []
update_times = []


for start in range(
    0,
    len(X_test_gpr),
    UPDATE_SIZE
):

    end = min(
        start + UPDATE_SIZE,
        len(X_test_gpr)
    )

    X_batch = X_test_gpr[start:end]
    y_batch = y_test.iloc[start:end].values

    batch_regimes = gmm.predict(X_batch)

    batch_pred = np.zeros(len(X_batch))
    batch_std = np.zeros(len(X_batch))


    # Predict before updating

    for r in range(N_REGIMES):

        idx = np.where(batch_regimes == r)[0]

        if len(idx) == 0:
            continue

        pred, std = adaptive_models[r].predict(
            X_batch[idx],
            return_std=True
        )

        batch_pred[idx] = pred
        batch_std[idx] = std


    rt_pred.extend(batch_pred)
    rt_actual.extend(y_batch)
    rt_std.extend(batch_std)


    # Add new sensor data

    affected = []

    for r in range(N_REGIMES):

        idx = np.where(batch_regimes == r)[0]

        if len(idx) == 0:
            continue

        affected.append(r)

        adaptive_X[r] = np.vstack([
            adaptive_X[r],
            X_batch[idx]
        ])[-WINDOW_SIZE:]

        adaptive_y[r] = np.concatenate([
            adaptive_y[r],
            y_batch[idx]
        ])[-WINDOW_SIZE:]


    # Fast refit using learned kernel

    start_time = time.time()

    for r in affected:

        adaptive_models[r] = realtime_gpr(
            kernels[r]
        )

        adaptive_models[r].fit(
            adaptive_X[r],
            adaptive_y[r]
        )

    update_times.append(
        time.time() - start_time
    )


# ------------------------------------------------------------
# 9. Real-time evaluation
# ------------------------------------------------------------

rt_pred = np.array(rt_pred)
rt_actual = np.array(rt_actual)
rt_std = np.array(rt_std)


rt_mae = mean_absolute_error(
    rt_actual,
    rt_pred
)

rt_rmse = np.sqrt(
    mean_squared_error(
        rt_actual,
        rt_pred
    )
)

rt_r2 = r2_score(
    rt_actual,
    rt_pred
)


# ------------------------------------------------------------
# 10. Static vs Real-Time comparison
# ------------------------------------------------------------

gpr_summary = pd.DataFrame({

    "Method": [
        "Static Mixture GPR",
        "Real-Time Mixture GPR"
    ],

    "MAE": [
        static_mae,
        rt_mae
    ],

    "RMSE": [
        static_rmse,
        rt_rmse
    ],

    "R²": [
        static_r2,
        rt_r2
    ],

    "Mean Uncertainty": [
        np.mean(static_std),
        np.mean(rt_std)
    ]

})


print("\n" + "=" * 65)
print("STATIC VS REAL-TIME MIXTURE GPR")
print("=" * 65)

display(
    gpr_summary.round(4)
)


print(
    "Average update time:",
    round(
        np.mean(update_times),
        4
    ),
    "seconds"
)


# ------------------------------------------------------------
# 11. Prediction visualisation
# ------------------------------------------------------------

plot_size = min(
    500,
    len(y_test)
)


plt.figure(figsize=(11, 5))

plt.plot(
    y_test.values[:plot_size],
    label="Actual"
)

plt.plot(
    static_pred[:plot_size],
    label="Static GPR"
)

plt.plot(
    rt_pred[:plot_size],
    label="Real-Time GPR"
)

plt.xlabel("Sensor Observation")
plt.ylabel("Moisture (VWC %)")

plt.title(
    "Mixture GPR — Static vs Real-Time"
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 12. Real-time uncertainty
# ------------------------------------------------------------

plt.figure(figsize=(11, 5))

plt.plot(
    rt_std
)

plt.xlabel("Sensor Observation")
plt.ylabel("Prediction Standard Deviation")

plt.title(
    "Mixture GPR — Real-Time Prediction Uncertainty"
)

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()


print(
    "\nModel 4 — Mixture GPR Real-Time completed."
)
```

## Model 4 — Mixture Gaussian Process Regression Real-Time

### Method — Initial Kernel Optimisation + Sliding-Window Fast Refitting

The Mixture GPR is first trained using the initial **2/3 training data**, where the Gaussian Process kernel parameters are optimised. The remaining **1/3 is used to simulate incoming sensor observations**.

Sensor data are processed in batches of **100 observations**. For each batch, the Gaussian Mixture Model identifies the corresponding regime and the relevant GPR predicts soil moisture and uncertainty.

After the actual sensor values become available, the new observations are added to the corresponding regime. A **sliding window of 1,000 observations per regime** is maintained, and the GPR is refitted using the previously learned kernel without repeating kernel optimisation.

**Initial GPR Training → Sensor Batch → Regime Identification → Prediction + Uncertainty → Sensor Feedback → Sliding-Window Update → Fast GPR Refit → Next Batch**

Performance is evaluated using **MAE, RMSE, R², prediction uncertainty, and update time**.

This highlights the importance of selecting an update strategy appropriate to each model rather than applying the same real-time learning approach to all algorithms.
