# Data final_semi_synthetic_winter_wheat_v2: Ver 2

## Part 1 — Load Data, Cleaning, EDA & Visualisation:

```python
# PART 1 — LOAD DATA, CLEANING, EDA & VISUALISATION
# 1. Import libraries for data processing, validation and visualisation.
from pathlib import Path
import csv
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
try:
    from IPython.display import display
except ImportError:
    # Support execution outside Jupyter or Google Colab.
    def display(value):
        if isinstance(value, (pd.DataFrame, pd.Series)):
            print(value.to_string())
        else:
            print(value)
# 2. Define the target and predictors used by the existing models.
TARGET = "semi_synthetic_moisture_vwc_pct"
FEATURES = [
    "longitudinal_distance_m",
    "lateral_distance_m",
    "depth_cm",
    "provided_moisture_vwc_pct",
]
# Identify a local spatial point independently of its depth.
POINT_COLUMNS = ["longitudinal_distance_m", "lateral_distance_m"]
# Preserve the chronological ordering required by Part 2.
SORT_COLUMNS = ["time_s", "lateral_distance_m", "depth_cm"]
# Define the expected configuration of the supplied dataset.
EXPECTED_DEPTHS = [5, 15]
EXPECTED_LATERALS = [0.0, 15.0, 30.0]
EXPECTED_SAMPLING_HZ = 10.0
EXPECTED_SPEED_KMH = 5.0
# The project allows depths up to 20 cm, but this dataset contains only 5 and 15 cm.
MAX_PROJECT_DEPTH_CM = 20.0
# Allow for coordinates rounded to the nearest 0.01 m.
COORD_ATOL_M = 0.005 + 1e-10
# Allow for rounding in the stored moisture fractions and percentages.
FRACTION_ATOL = 0.5e-6 + 0.5e-8 + 1e-12
# Define tolerances for depth multipliers and elapsed time.
MULTIPLIER_ATOL = 1e-6 + 1e-12
TIME_ATOL_S = 1e-8
# Full-dataset EDA is descriptive; do not use test-set patterns for model tuning.
# 3. Load the dataset and preserve an unchanged raw copy.
DATA_PATH = Path("final_semi_synthetic_winter_wheat_v2.csv")
# Support uploaded filenames containing suffixes such as "(7)".
if not DATA_PATH.is_file():
    candidates = sorted({
        path.resolve()
        for folder in (Path("."), Path("upload"))
        for path in folder.glob("final_semi_synthetic_winter_wheat_v2*.csv")
    })
    # Avoid silently selecting a file when several dataset copies exist.
    if len(candidates) != 1:
        raise FileNotFoundError(
            "Set DATA_PATH explicitly. "
            f"Found {len(candidates)} candidate CSV files: {candidates}"
        )
    DATA_PATH = candidates[0]
# Inspect the original header before pandas can rename duplicate column names.
with DATA_PATH.open(newline="", encoding="utf-8-sig") as source:
    header = next(csv.reader(source))
if len(header) != len(set(header)):
    raise ValueError("Duplicate column names in the original CSV header.")
df_raw = pd.read_csv(DATA_PATH)
print("Loaded file:", DATA_PATH.name)
print("Raw dataset shape:", df_raw.shape)
print("\nFirst 5 rows:")
display(df_raw.head())
# 4. Define reusable functions for recording and checking audit results.
audit_rows = []
def record_check(name, condition, detail, review=False):
    # PASS indicates that the specified rule passed.
    # REVIEW identifies a limitation or a logged cleaning action.
    # FAIL identifies an inconsistency that must be resolved.
    status = "PASS" if bool(condition) else ("REVIEW" if review else "FAIL")
    audit_rows.append({
        "check": name,
        "status": status,
        "detail": str(detail),
    })
def show_audit_and_stop_if_failed():
    # Display the accumulated checks and stop if a critical check fails.
    report = pd.DataFrame(audit_rows)
    display(report)
    if (report["status"] == "FAIL").any():
        failed = report.loc[report["status"] == "FAIL", "check"].tolist()
        raise ValueError(
            "Resolve failed data checks before continuing: " + "; ".join(failed)
        )
# 5. Validate the expected dataset schema.
NUMERIC_COLUMNS = [
    "sample_index",
    "time_s",
    "longitudinal_distance_m",
    "lateral_distance_m",
    "provided_moisture_vwc_pct",
    "provided_moisture_fraction",
    "depth_cm",
    "depth_multiplier",
    TARGET,
    "semi_synthetic_moisture_fraction",
    "sampling_rate_hz",
    "tractor_speed_kmh",
    "reference_ratio_median_15_to_5",
]
METADATA_COLUMNS = [
    "reference_location",
    "reference_window_start",
    "reference_window_end",
    "spatial_source",
    "depth_profile_reference",
    "data_type",
]
missing_columns = sorted(
    set(NUMERIC_COLUMNS + METADATA_COLUMNS) - set(df_raw.columns)
)
record_check(
    "Required schema",
    not missing_columns,
    f"Missing columns: {missing_columns}",
)
record_check(
    "Non-empty dataset",
    len(df_raw) > 0,
    f"Rows: {len(df_raw):,}",
)
if missing_columns or df_raw.empty:
    show_audit_and_stop_if_failed()
# Work on a separate copy so the original data remain available for inspection.
df = df_raw.copy(deep=True)
# 6. Check missing values, blank strings, invalid numbers and infinity.
quality_rows = []
for column in df.columns:
    original = df[column]
    # Detect empty strings and strings containing only whitespace.
    blank_strings = int(
        original.astype("string").str.strip().eq("").fillna(False).sum()
    )
    invalid_numeric = 0
    infinite_values = 0
    if column in NUMERIC_COLUMNS:
        # Convert expected numeric columns and explicitly record conversion failures.
        converted = pd.to_numeric(original, errors="coerce")
        invalid_numeric = int((original.notna() & converted.isna()).sum())
        infinite_values = int(
            np.isinf(converted.to_numpy(dtype=float)).sum()
        )
        df[column] = converted
    quality_rows.append({
        "column": column,
        "raw_dtype": str(original.dtype),
        "missing": int(original.isna().sum()),
        "blank_strings": blank_strings,
        "invalid_numeric": invalid_numeric,
        "infinite": infinite_values,
        "unique_values": int(original.nunique(dropna=True)),
    })
quality_report = pd.DataFrame(quality_rows)
print("\nData quality by column:")
display(quality_report)
for field in ["missing", "blank_strings", "invalid_numeric", "infinite"]:
    count = int(quality_report[field].sum())
    record_check(
        f"No {field}",
        count == 0,
        f"Affected cells: {count}",
    )
# Do not continue with missing, invalid or infinite values.
if any(row["status"] == "FAIL" for row in audit_rows):
    show_audit_and_stop_if_failed()
# 7. Check duplicates and perform controlled cleaning.
# sample_index repeats across lateral positions and depths, so it is not unique alone.
raw_duplicates = int(df_raw.duplicated().sum())
record_check(
    "No exact duplicate raw rows",
    raw_duplicates == 0,
    f"Exact duplicate rows removed: {raw_duplicates}",
    review=True,
)
# Remove only rows that are exact duplicates in the original dataset.
df = df.loc[~df_raw.duplicated()].copy()
# Detect duplicates introduced by equivalent numeric representations.
record_check(
    "No duplicate rows after numeric conversion",
    not df.duplicated().any(),
    f"Remaining duplicates: {df.duplicated().sum()}",
)
# Check alternative observation keys for conflicting records.
observation_keys = [
    ["sample_index", "lateral_distance_m", "depth_cm"],
    SORT_COLUMNS,
    POINT_COLUMNS + ["depth_cm"],
]
for keys in observation_keys:
    duplicates = df.duplicated(keys, keep=False)
    record_check(
        "Unique key: " + ", ".join(keys),
        not duplicates.any(),
        f"Rows with repeated keys: {int(duplicates.sum())}",
    )
    if duplicates.any():
        print("\nConflicting observation keys:", keys)
        display(df.loc[duplicates].head(10))
if any(row["status"] == "FAIL" for row in audit_rows):
    show_audit_and_stop_if_failed()
# Sort records without changing their measurements or removing statistical outliers.
df = df.sort_values(SORT_COLUMNS, kind="stable").reset_index(drop=True)
print("\nDataset shape after cleaning:", df.shape)
print(
    "Cleaning: remove exact duplicate rows and sort; "
    "no imputation, clipping or outlier deletion."
)
print("\nDataset information:")
df.info()
print("\nNumeric summary statistics:")
display(df[NUMERIC_COLUMNS].describe().T)
# Identify constant fields that describe the dataset configuration or provenance.
constant_columns = df.columns[df.nunique(dropna=False).eq(1)].tolist()
print("\nConstant columns:", constant_columns)
display(
    df[constant_columns].iloc[[0]].T.rename(columns={0: "constant_value"})
)
# 8. Validate moisture ranges and percentage-to-fraction conversions.
for column in ["provided_moisture_vwc_pct", TARGET]:
    record_check(
        f"{column}: 0–100 VWC %",
        df[column].between(0, 100).all(),
        f"Observed range: {df[column].min():.6f}–{df[column].max():.6f}",
    )
for column in [
    "provided_moisture_fraction",
    "semi_synthetic_moisture_fraction",
]:
    record_check(
        f"{column}: 0–1",
        df[column].between(0, 1).all(),
        f"Observed range: {df[column].min():.6f}–{df[column].max():.6f}",
    )
moisture_unit_pairs = [
    ("provided_moisture_vwc_pct", "provided_moisture_fraction"),
    (TARGET, "semi_synthetic_moisture_fraction"),
]
for percent, fraction in moisture_unit_pairs:
    # A moisture fraction should equal its percentage value divided by 100.
    error = (df[percent] / 100 - df[fraction]).abs()
    record_check(
        f"Percent/fraction agreement: {fraction}",
        (error <= FRACTION_ATOL).all(),
        f"Max absolute fraction error: {error.max():.9g}; "
        f"tolerance: {FRACTION_ATOL:.9g}",
    )
# 9. Validate depth, lateral positions, sample indices and operating parameters.
record_check(
    "Expected depths",
    sorted(df.depth_cm.unique()) == EXPECTED_DEPTHS,
    f"Observed: {sorted(df.depth_cm.unique())}; expected: {EXPECTED_DEPTHS}",
)
record_check(
    "Project depth bounds",
    df.depth_cm.between(0, MAX_PROJECT_DEPTH_CM).all(),
    f"Allowed project range: 0–{MAX_PROJECT_DEPTH_CM:g} cm",
)
record_check(
    "Expected lateral positions",
    sorted(df.lateral_distance_m.unique()) == EXPECTED_LATERALS,
    f"Observed: {sorted(df.lateral_distance_m.unique())}",
)
for column in [
    "time_s",
    "longitudinal_distance_m",
    "lateral_distance_m",
    "sample_index",
]:
    record_check(
        f"Nonnegative {column}",
        df[column].ge(0).all(),
        f"Minimum: {df[column].min():g}",
    )
record_check(
    "Integer sample indices",
    np.equal(df.sample_index, np.floor(df.sample_index)).all(),
    f"Index range: {df.sample_index.min():g}–{df.sample_index.max():g}",
)
for column, expected in [
    ("sampling_rate_hz", EXPECTED_SAMPLING_HZ),
    ("tractor_speed_kmh", EXPECTED_SPEED_KMH),
]:
    record_check(
        f"Expected {column}",
        np.isclose(df[column], expected, rtol=0, atol=1e-10).all(),
        f"Observed: {sorted(df[column].unique())}; expected: {expected:g}",
    )
# 10. Check reference metadata and the source date window.
for column in METADATA_COLUMNS + ["reference_ratio_median_15_to_5"]:
    record_check(
        f"Single reference value: {column}",
        df[column].nunique() == 1,
        f"Unique values: {df[column].nunique()}",
    )
for column in ["depth_multiplier", "reference_ratio_median_15_to_5"]:
    record_check(
        f"Positive {column}",
        df[column].gt(0).all(),
        f"Minimum: {df[column].min():.6f}",
    )
reference_start = pd.to_datetime(
    df.reference_window_start,
    format="%Y-%m-%d",
    errors="coerce",
)
reference_end = pd.to_datetime(
    df.reference_window_end,
    format="%Y-%m-%d",
    errors="coerce",
)
record_check(
    "Valid ordered reference dates",
    (
        reference_start.notna()
        & reference_end.notna()
        & reference_start.le(reference_end)
    ).all(),
    f"Reference window: {df.reference_window_start.iloc[0]} "
    f"to {df.reference_window_end.iloc[0]}",
)
# time_s is elapsed replay time; the reference dates describe the source window.
show_audit_and_stop_if_failed()
# 11. Check that each local point contains a consistent pair of depths.
point_groups_eda = df.groupby(POINT_COLUMNS, sort=False)
point_structure = point_groups_eda.agg(
    rows=(TARGET, "size"),
    depths=("depth_cm", "nunique"),
    times=("time_s", "nunique"),
    samples=("sample_index", "nunique"),
    input_values=("provided_moisture_vwc_pct", "nunique"),
)
record_check(
    "Paired depths per local point",
    (
        point_structure["rows"].eq(2)
        & point_structure["depths"].eq(2)
    ).all(),
    f"Local points: {len(point_structure):,}; expected rows/point: 2",
)
record_check(
    "Shared time, sample and input at paired depths",
    point_structure[["times", "samples", "input_values"]].eq(1).all().all(),
    "The two depth rows must describe the same source point.",
)
# Each timestamp should contain three lateral positions multiplied by two depths.
expected_grid = pd.MultiIndex.from_product(
    [
        np.sort(df.time_s.unique()),
        EXPECTED_LATERALS,
        EXPECTED_DEPTHS,
    ],
    names=SORT_COLUMNS,
)
observed_grid = pd.MultiIndex.from_frame(df[SORT_COLUMNS])
missing_grid = expected_grid.difference(observed_grid)
extra_grid = observed_grid.difference(expected_grid)
record_check(
    "Complete time/lateral/depth grid",
    len(missing_grid) == len(extra_grid) == 0,
    f"Missing combinations: {len(missing_grid)}; unexpected: {len(extra_grid)}",
)
# 12. Check sampling intervals and movement within each lateral/depth stream.
# Checking intervals across the complete table would include repeated timestamps.
stream_rows = []
for (lateral, depth), stream in df.groupby(
    ["lateral_distance_m", "depth_cm"]
):
    stream = stream.sort_values("time_s")
    dt = np.diff(stream.time_s.to_numpy())
    dx = np.diff(stream.longitudinal_distance_m.to_numpy())
    expected_dt = 1 / stream.sampling_rate_hz.iloc[0]
    # Convert tractor speed from km/h to m/s before calculating distance.
    expected_dx = stream.tractor_speed_kmh.iloc[0] / 3.6 * dt
    stream_rows.append({
        "lateral_m": lateral,
        "depth_cm": depth,
        "rows": len(stream),
        "min_dt_s": dt.min() if dt.size else np.nan,
        "max_dt_s": dt.max() if dt.size else np.nan,
        "mean_dx_m": dx.mean() if dx.size else np.nan,
        "min_dx_m": dx.min() if dx.size else np.nan,
        "max_dx_m": dx.max() if dx.size else np.nan,
    })
    record_check(
        f"10 Hz intervals: lateral {lateral:g}, depth {depth:g}",
        (
            dt.size > 0
            and np.isclose(
                dt, expected_dt, rtol=0, atol=TIME_ATOL_S
            ).all()
        ),
        f"Expected interval: {expected_dt:.3f} s",
    )
    record_check(
        f"Contiguous indices: lateral {lateral:g}, depth {depth:g}",
        np.array_equal(
            stream.sample_index.to_numpy(),
            np.arange(len(stream)),
        ),
        "Indices must start at zero and increase by one.",
    )
    # Account for rounding at both endpoints of a distance interval.
    record_check(
        f"Distance increments: lateral {lateral:g}, depth {depth:g}",
        (
            dx.size > 0
            and (dx > 0).all()
            and (np.abs(dx - expected_dx) <= 2 * COORD_ATOL_M).all()
        ),
        "Tolerance accounts for rounding both interval endpoints.",
    )
stream_summary = pd.DataFrame(stream_rows)
print("\nSampling and spatial coverage by stream:")
display(stream_summary)
record_check(
    "Equal coverage across lateral/depth streams",
    stream_summary.rows.nunique() == 1,
    f"Counts: {stream_summary.rows.tolist()}",
)
# Independently reconstruct elapsed time from sample index and sampling rate.
time_error = (
    df.time_s - df.sample_index / df.sampling_rate_hz
).abs()
# Independently reconstruct longitudinal position from speed and elapsed time.
position_error = (
    df.longitudinal_distance_m
    - df.time_s * df.tractor_speed_kmh / 3.6
).abs()
record_check(
    "time_s = sample_index / sampling_rate_hz",
    (time_error <= TIME_ATOL_S).all(),
    f"Max time error: {time_error.max():.9g} s",
)
record_check(
    "distance = speed × time",
    (position_error <= COORD_ATOL_M).all(),
    f"Max position error: {position_error.max():.9g} m; "
    f"tolerance: {COORD_ATOL_M:.9g} m",
)
print(f"\nUnique replay timestamps: {df.time_s.nunique():,}")
print(f"Unique spatial points: {len(point_structure):,}")
print(f"Time range: {df.time_s.min():g}–{df.time_s.max():g} s")
print(
    f"Longitudinal range: {df.longitudinal_distance_m.min():g}–"
    f"{df.longitudinal_distance_m.max():g} m"
)
print(
    "Expected sampling distance:",
    f"{EXPECTED_SPEED_KMH / 3.6 / EXPECTED_SAMPLING_HZ:.6f} m",
)
if any(row["status"] == "FAIL" for row in audit_rows):
    show_audit_and_stop_if_failed()
# 13. Verify the semi-synthetic target-generation relationship.
reconstructed_target = (
    df.provided_moisture_vwc_pct * df.depth_multiplier
)
formula_error = (df[TARGET] - reconstructed_target).abs()
# Propagate half-unit rounding bounds for the stored multiplier and target.
# Provided moisture is treated as the stored input value.
formula_atol = (
    df.provided_moisture_vwc_pct.abs() * 0.5e-6
    + 0.5e-6
    + 1e-12
)
record_check(
    "Target construction within CSV precision",
    (formula_error <= formula_atol).all(),
    f"Max error: {formula_error.max():.9f} VWC percentage points",
)
multiplier_summary = (
    df.groupby("depth_cm")["depth_multiplier"]
    .agg(["nunique", "min", "max"])
)
record_check(
    "One multiplier per depth",
    multiplier_summary["nunique"].eq(1).all(),
    f"Unique multiplier counts: {multiplier_summary['nunique'].to_dict()}",
)
ratio = float(df.reference_ratio_median_15_to_5.iloc[0])
# Check the implied normalised construction: a5 = 2/(1+r), a15 = 2r/(1+r).
# This checks internal consistency and does not independently verify calibration.
implied_multipliers = {
    5: 2 / (1 + ratio),
    15: 2 * ratio / (1 + ratio),
}
multiplier_summary["implied_from_ratio"] = (
    multiplier_summary.index.map(implied_multipliers)
)
multiplier_summary["absolute_difference"] = (
    multiplier_summary["min"]
    - multiplier_summary["implied_from_ratio"]
).abs()
record_check(
    "Multipliers agree with normalised reference ratio",
    multiplier_summary["absolute_difference"].le(MULTIPLIER_ATOL).all(),
    "Max multiplier difference: "
    f"{multiplier_summary['absolute_difference'].max():.9g}",
)
print("\nDepth multiplier consistency:")
display(multiplier_summary)
# Compare the target values at both depths for each spatial point.
paired_target = df.pivot(
    index=POINT_COLUMNS,
    columns="depth_cm",
    values=TARGET,
)
# Exclude zero denominators when calculating depth ratios.
observed_ratio = paired_target[15].div(
    paired_target[5].replace(0, np.nan)
)
print("\nReference ratio (15/5 cm):", ratio)
print("Observed target ratio for nonzero 5 cm targets:")
display(observed_ratio.describe().to_frame("15_to_5_target_ratio"))
print(
    "\nMaximum target-construction difference:",
    f"{formula_error.max():.9f}",
)
print("Formula reconstruction MAE:", f"{formula_error.mean():.9g}")
print(
    "Formula reconstruction RMSE:",
    f"{np.sqrt(np.mean(formula_error ** 2)):.9g}",
)
print(
    "These are construction diagnostics, not fitted-model "
    "or held-out evaluation metrics."
)
# Ensure that the target and its fraction representation are not model predictors.
record_check(
    "No direct target copy in FEATURES",
    (
        TARGET not in FEATURES
        and "semi_synthetic_moisture_fraction" not in FEATURES
    ),
    f"Predictors: {FEATURES}",
)
# Provided moisture and depth still almost completely determine this target.
# High model scores therefore require interpretation within this construction.
if any(row["status"] == "FAIL" for row in audit_rows):
    show_audit_and_stop_if_failed()
# 14. Summarise moisture by lateral position and depth.
moisture_summary = (
    df.groupby(["lateral_distance_m", "depth_cm"])[TARGET]
    .agg(
        count="size",
        mean="mean",
        std="std",
        minimum="min",
        q25=lambda values: values.quantile(0.25),
        median="median",
        q75=lambda values: values.quantile(0.75),
        maximum="max",
    )
    .reset_index()
)
print("\nMoisture summary by lateral position and depth:")
display(moisture_summary)
# 15. Flag potential outliers using IQR within each lateral/depth group.
# Group-specific fences avoid treating the imposed depth difference as an outlier.
outlier_rows = []
for (lateral, depth), group in df.groupby(
    ["lateral_distance_m", "depth_cm"]
):
    values = group[TARGET]
    q1, q3 = values.quantile([0.25, 0.75])
    iqr = q3 - q1
    lower_fence = q1 - 1.5 * iqr
    upper_fence = q3 + 1.5 * iqr
    flagged = (values < lower_fence) | (values > upper_fence)
    outlier_rows.append({
        "lateral_m": lateral,
        "depth_cm": depth,
        "lower_fence": lower_fence,
        "upper_fence": upper_fence,
        "flagged_rows": int(flagged.sum()),
        "flagged_pct": 100 * flagged.mean(),
    })
outlier_summary = pd.DataFrame(outlier_rows)
print("\nPotential outliers using within-group IQR:")
display(outlier_summary)
print(
    "IQR flags are retained for review; "
    "no statistical outliers are automatically removed."
)
# Calculate correlations for the selected predictors and target only.
correlation = df[FEATURES + [TARGET]].corr()
print("\nPearson correlation:")
display(correlation)
# 16. Preview the existing grouped train–test split without changing Part 2.
# Both depth observations from each local point remain in the same partition.
eda_points = (
    df[POINT_COLUMNS + ["time_s"]]
    .drop_duplicates(POINT_COLUMNS)
    .sort_values(["time_s", "lateral_distance_m"])
)
eda_split_index = 2 * len(eda_points) // 3
eda_training_keys = pd.MultiIndex.from_frame(
    eda_points.iloc[:eda_split_index][POINT_COLUMNS]
)
eda_train_mask = pd.MultiIndex.from_frame(
    df[POINT_COLUMNS]
).isin(eda_training_keys)
eda_train = df.loc[eda_train_mask]
eda_test = df.loc[~eda_train_mask]
# Identify timestamps shared across different spatial points at the split boundary.
shared_times = sorted(
    set(eda_train.time_s) & set(eda_test.time_s)
)
train_point_keys = set(
    map(tuple, eda_train[POINT_COLUMNS].to_numpy())
)
test_point_keys = set(
    map(tuple, eda_test[POINT_COLUMNS].to_numpy())
)
record_check(
    "Existing split has disjoint local points",
    train_point_keys.isdisjoint(test_point_keys),
    "Both depths of each spatial point remain in the same partition.",
)
record_check(
    "Existing split is nondecreasing in time",
    eda_train.time_s.max() <= eda_test.time_s.min(),
    f"Train ends: {eda_train.time_s.max():g} s; "
    f"test starts: {eda_test.time_s.min():g} s",
)
record_check(
    "Existing split has disjoint timestamps",
    len(shared_times) == 0,
    f"Shared boundary timestamps: {shared_times}",
    review=True,
)
split_summary = pd.DataFrame([
    {
        "partition": label,
        "rows": len(part),
        "points": len(part[POINT_COLUMNS].drop_duplicates()),
        "timestamps": part.time_s.nunique(),
        "start_s": part.time_s.min(),
        "end_s": part.time_s.max(),
        "target_mean": part[TARGET].mean(),
        "target_std": part[TARGET].std(),
    }
    for label, part in [
        ("Train preview", eda_train),
        ("Test preview", eda_test),
    ]
])
print("\nExisting train–test split preview:")
display(split_summary)
# Compare test ranges with training ranges to describe extrapolation.
coverage_rows = []
for column in FEATURES + [TARGET]:
    train_min = eda_train[column].min()
    train_max = eda_train[column].max()
    outside_train_range = ~eda_test[column].between(train_min, train_max)
    coverage_rows.append({
        "variable": column,
        "train_min": train_min,
        "train_max": train_max,
        "test_min": eda_test[column].min(),
        "test_max": eda_test[column].max(),
        "test_outside_train_range_pct": 100 * outside_train_range.mean(),
    })
print("\nTrain–test range comparison:")
display(pd.DataFrame(coverage_rows))
print(
    "Ranges are descriptive. Longitudinal extrapolation "
    "is expected under this ordered split."
)
# Reproduce the existing batch-count calculation using 50 timestamps per batch.
eda_test_ticks = np.sort(eda_test.time_s.unique())
eda_batch_count = (len(eda_test_ticks) + 49) // 50
print("\nExisting 50-timestamp batches:", eda_batch_count)
print("Updates affecting future predictions:", eda_batch_count - 1)
# Strict future-time evaluation would require grouping entire timestamps together.
# Any change to that split requires rerunning the model comparisons.
# 17. Configure consistent plot colours and formatting.
depth_colors = {5: "#287C8E", 15: "#CB7438"}
lane_colors = {0.0: "#287C8E", 15.0: "#8061A5", 30.0: "#CB7438"}
plt.rcParams.update({
    "font.size": 10,
    "axes.spines.top": False,
    "axes.spines.right": False,
})
# 18. Visualise distributions, depth differences and feature relationships.
fig, axes = plt.subplots(
    2, 2, figsize=(13, 9), constrained_layout=True
)
# Use shared bin boundaries when comparing distributions.
bins = np.linspace(df[TARGET].min(), df[TARGET].max(), 40)
# Plot target distributions separately for each depth.
for depth in EXPECTED_DEPTHS:
    axes[0, 0].hist(
        df.loc[df.depth_cm == depth, TARGET],
        bins=bins,
        alpha=0.62,
        color=depth_colors[depth],
        label=f"{depth} cm",
    )
axes[0, 0].set(
    title="Target distribution by depth",
    xlabel="Moisture (VWC %)",
    ylabel="Observations",
)
axes[0, 0].legend()
# Compare moisture distributions across all lateral/depth combinations.
box_data = [
    df.loc[
        (df.lateral_distance_m == lateral) & (df.depth_cm == depth),
        TARGET,
    ]
    for lateral in EXPECTED_LATERALS
    for depth in EXPECTED_DEPTHS
]
axes[0, 1].boxplot(box_data, showfliers=True)
axes[0, 1].set_xticks(range(1, 7))
axes[0, 1].set_xticklabels([
    f"{lateral:g} m\n{depth} cm"
    for lateral in EXPECTED_LATERALS
    for depth in EXPECTED_DEPTHS
])
axes[0, 1].set(
    title="Moisture by lateral position and depth",
    xlabel="Lateral position / depth",
    ylabel="Moisture (VWC %)",
)
# Show the input–target relationship produced by the depth multipliers.
for depth in EXPECTED_DEPTHS:
    subset = df.loc[df.depth_cm == depth]
    axes[1, 0].scatter(
        subset.provided_moisture_vwc_pct,
        subset[TARGET],
        s=8,
        alpha=0.25,
        color=depth_colors[depth],
        label=f"{depth} cm",
    )
axes[1, 0].set(
    title="Input–target relationship imposed by construction",
    xlabel="Provided moisture (VWC %)",
    ylabel="Target moisture (VWC %)",
)
axes[1, 0].legend()
# Display Pearson correlations; correlation does not establish causation.
short_labels = ["Longitudinal", "Lateral", "Depth", "Provided", "Target"]
image = axes[1, 1].imshow(
    correlation, vmin=-1, vmax=1, cmap="RdBu_r"
)
axes[1, 1].set_xticks(range(5))
axes[1, 1].set_yticks(range(5))
axes[1, 1].set_xticklabels(short_labels, rotation=30, ha="right")
axes[1, 1].set_yticklabels(short_labels)
# Annotate each heatmap cell with its correlation coefficient.
for i in range(5):
    for j in range(5):
        value = correlation.iloc[i, j]
        axes[1, 1].text(
            j,
            i,
            f"{value:.2f}",
            ha="center",
            va="center",
            color="white" if abs(value) > 0.6 else "black",
        )
axes[1, 1].set_title("Pearson correlation: predictors and target")
fig.colorbar(image, ax=axes[1, 1], shrink=0.8, label="Pearson r")
plt.show()
# 19. Visualise longitudinal moisture profiles and maps at both depths.
fig, axes = plt.subplots(
    2, 2, figsize=(14, 9), constrained_layout=True
)
# Use common moisture limits to support comparison between depths.
vmin = df[TARGET].min()
vmax = df[TARGET].max()
for row, depth in enumerate(EXPECTED_DEPTHS):
    subset = df.loc[df.depth_cm == depth]
    for lateral in EXPECTED_LATERALS:
        lane = (
            subset.loc[subset.lateral_distance_m == lateral]
            .sort_values("longitudinal_distance_m")
        )
        # Display raw measurements with low opacity.
        axes[row, 0].plot(
            lane.longitudinal_distance_m,
            lane[TARGET],
            color=lane_colors[lateral],
            alpha=0.1,
            linewidth=0.5,
        )
        # Add a trailing mean for visual interpretation only.
        # These smoothed values are not supplied to the models.
        axes[row, 0].plot(
            lane.longitudinal_distance_m,
            lane[TARGET].rolling(50, min_periods=1).mean(),
            color=lane_colors[lateral],
            linewidth=1.4,
            label=f"Lateral {lateral:g} m",
        )
    axes[row, 0].set(
        title=f"{depth} cm: raw profiles and trailing 50-point mean",
        xlabel="Longitudinal distance (m)",
        ylabel="Moisture (VWC %)",
        ylim=(vmin - 0.2, vmax + 0.2),
    )
    axes[row, 0].legend(fontsize=8, ncol=3)
    # Map only the sampled tracks without interpolating between lateral positions.
    scatter = axes[row, 1].scatter(
        subset.longitudinal_distance_m,
        subset.lateral_distance_m,
        c=subset[TARGET],
        cmap="viridis",
        vmin=vmin,
        vmax=vmax,
        s=12,
    )
    axes[row, 1].set(
        title=f"Local moisture map: {depth} cm",
        xlabel="Longitudinal distance (m)",
        ylabel="Lateral distance (m)",
        yticks=EXPECTED_LATERALS,
        ylim=(-4, 34),
    )
# A shared colour scale makes moisture values comparable across both maps.
fig.colorbar(
    scatter,
    ax=axes[:, 1],
    label="Moisture (VWC %), shared scale",
    shrink=0.9,
)
plt.show()
# 20. Compare train–test distributions and inspect moisture over replay time.
fig, axes = plt.subplots(
    1, 2, figsize=(13, 4.5), constrained_layout=True
)
# Use density because the train and test partitions contain different row counts.
for label, part, color in [
    ("Train preview", eda_train, "#287C8E"),
    ("Test preview", eda_test, "#CB7438"),
]:
    axes[0].hist(
        part[TARGET],
        bins=bins,
        density=True,
        alpha=0.55,
        color=color,
        label=label,
    )
axes[0].set(
    title="Target distribution across the existing split",
    xlabel="Moisture (VWC %)",
    ylabel="Density",
)
axes[0].legend()
# Average across lateral tracks at each timestamp before applying a trailing mean.
for depth in EXPECTED_DEPTHS:
    series = (
        df.loc[df.depth_cm == depth]
        .groupby("time_s")[TARGET]
        .mean()
    )
    axes[1].plot(
        series.index,
        series.rolling(50, min_periods=1).mean(),
        color=depth_colors[depth],
        label=f"{depth} cm",
    )
# Mark the beginning of the test partition and identify a shared boundary if present.
boundary_label = (
    "Test begins (shared boundary)" if shared_times else "Test begins"
)
axes[1].axvline(
    eda_test.time_s.min(),
    color="#555555",
    linestyle="--",
    label=boundary_label,
)
axes[1].set(
    title="Mean across lateral tracks: trailing 50 timestamps",
    xlabel="Replay time (s)",
    ylabel="Moisture (VWC %)",
)
axes[1].legend(fontsize=8)
plt.show()
# 21. Display the complete data quality and parameter consistency report.
audit_report = pd.DataFrame(audit_rows)
print("\nFINAL DATA QUALITY AND PARAMETER CONSISTENCY REPORT")
# Show all audit rows and avoid truncating their explanations.
with pd.option_context(
    "display.max_rows", None,
    "display.max_colwidth", None,
):
    display(audit_report)
print("\nStatus counts:", audit_report.status.value_counts().to_dict())
# Stop before modelling if any critical inconsistency remains.
if (audit_report.status == "FAIL").any():
    raise ValueError(
        "Audit failed. Resolve FAIL items before running Part 2."
    )
print(
    f"\nValidated modelling frame: "
    f"{len(df):,} rows × {len(df.columns)} columns."
)
print("Predictors:", FEATURES)
print("Target:", TARGET)
print("\nRemaining reviews:")
display(audit_report.loc[audit_report.status == "REVIEW"])
print(
    "\nInterpretation: the target is semi-synthetic and is "
    "almost completely determined by provided moisture "
    "and the depth multiplier."
)
print(
    "Confirm that provided moisture is available at inference "
    "time when interpreting model performance."
)
# Internal consistency does not independently verify sensor accuracy.
# Sampling at 10 Hz does not verify the actuator's seed-release capacity.
# Sensor-to-gate distance is absent, so tau = d/v cannot be validated here.
# longitudinal_distance_m represents field position, not the sensor-to-gate offset.
# Only the 5 cm and 15 cm depth levels are represented in this dataset.
# The variables df, TARGET and FEATURES remain available for Part 2.
# This section performs EDA and validation without training the six models.
```

### 2.2. Evaluation Metrics:

| Metric | Interpretation | Preferred result |
|---|---|---|
| **Mean Absolute Error (MAE)** | Average absolute difference between predicted and reference moisture. | Lower |
| **Root Mean Squared Error (RMSE)** | Measures prediction error while giving greater weight to large errors. | Lower |
| **Coefficient of Determination (R²)** | Measures how well predictions explain variation in reference moisture. Values can be negative, with a maximum of 1. | Higher |
| **Bias** | Average of predicted moisture minus reference moisture. Positive values indicate overprediction; negative values indicate underprediction. | Closer to 0 |
| **Prediction Latency** | Median execution time of individual single-sample prediction calls, including preprocessing and excluding warm-up calls. | Lower |
| **Average Update Time** | Mean execution time required to update the model after receiving a labelled sensor batch. | Lower |

**MAE, RMSE and Bias** are expressed in **VWC percentage points**. **Prediction Latency** is reported in **milliseconds per prediction**, while **Average Update Time** is reported in **seconds per update**. Timing measurements exclude sensor communication and actuator delays.

### 2.3. Shared Evaluation and Real-Time Updating Procedure:

```python
# SHARED SETUP — METRICS AND SENSOR-STREAM EVALUATION
import copy
import warnings
from time import perf_counter
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from scipy.optimize import minimize
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.compose import TransformedTargetRegressor
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
from sklearn.neural_network import MLPRegressor
from sklearn.mixture import GaussianMixture
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import ConstantKernel, RBF, WhiteKernel
from threadpoolctl import threadpool_limits
from IPython.display import display
# 1. Configuration and existing train–test partitions
SEED = 42
THREAD_LIMITER = threadpool_limits(limits=1)
np.random.seed(SEED)
FEATURES = ["longitudinal_distance_m", "lateral_distance_m", "depth_cm", "provided_moisture_vwc_pct"]
TARGET = "semi_synthetic_moisture_vwc_pct"
# Keep the partitions already created in Section 2.1.
sort_columns = ["time_s", "lateral_distance_m", "depth_cm"]
train_df = train_df.sort_values(sort_columns).reset_index(drop=True)
test_df = test_df.sort_values(sort_columns).reset_index(drop=True)
assert train_df.time_s.max() <= test_df.time_s.min()
point_columns = ["longitudinal_distance_m", "lateral_distance_m"]
assert set(map(tuple, train_df[point_columns].to_numpy())).isdisjoint(set(map(tuple, test_df[point_columns].to_numpy())))
X_train = train_df[FEATURES].to_numpy(dtype=float)
y_train = train_df[TARGET].to_numpy(dtype=float)
X_test = test_df[FEATURES].to_numpy(dtype=float)
y_test = test_df[TARGET].to_numpy(dtype=float)
# 2. Training-only validation and sensor batches
training_times = np.sort(train_df.time_s.unique())
validation_start = training_times[int(0.85 * len(training_times))]
INITIAL_FIT_MASK = train_df.time_s.to_numpy() < validation_start
TICKS_PER_BATCH = 50
test_times = np.sort(test_df.time_s.unique())
BATCHES = [np.flatnonzero(test_df.time_s.isin(test_times[i:i + TICKS_PER_BATCH])) for i in range(0, len(test_times), TICKS_PER_BATCH)]
RESULTS = []
PREDICTIONS = {}
BATCH_LOGS = {}
# 3. Prediction quality metrics
def regression_metrics(actual, predicted):
    actual = np.asarray(actual, dtype=float).reshape(-1)
    predicted = np.asarray(predicted, dtype=float).reshape(-1)
    if actual.shape != predicted.shape or not actual.size:
        raise ValueError("Targets and predictions must have matching lengths.")
    if not (np.isfinite(actual).all() and np.isfinite(predicted).all()):
        raise ValueError("Non-finite target or prediction.")
    return {
        "MAE": mean_absolute_error(actual, predicted),
        "RMSE": np.sqrt(mean_squared_error(actual, predicted)),
        "R2": r2_score(actual, predicted),
        "Bias": np.mean(predicted - actual),
    }
# 4. Single-sample prediction latency
def single_sample_latencies(model, X):
    # Warm-up is excluded from timing.
    model.predict(X[:1])
    timings = []
    selected_rows = np.linspace(0, len(X) - 1, min(10, len(X)), dtype=int)
    for i in selected_rows:
        started = perf_counter()
        model.predict(X[i:i + 1])
        timings.append(1000 * (perf_counter() - started))
    return timings
# 5. Shared wrapper for periodic model refitting
class PeriodicRefitRegressor:
    def __init__(self, factory, window=None):
        self.factory = factory
        self.window = window
    def fit(self, X, y):
        self.model = self.factory().fit(X, y)
        self.X = X.copy()
        self.y = y.copy()
        if self.window is not None:
            self.X = self.X[-self.window:]
            self.y = self.y[-self.window:]
        return self
    def predict(self, X):
        return self.model.predict(X)
    def update(self, X, y):
        self.X = np.vstack([self.X, X])
        self.y = np.concatenate([self.y, y])
        if self.window is not None:
            self.X = self.X[-self.window:]
            self.y = self.y[-self.window:]
        self.model = self.factory().fit(self.X, self.y)
# 6. Shared Static versus Real-Time evaluation
def evaluate_model(name, model, update_method):
    global RESULTS
    print(f"\nTraining {name}...", flush=True)
    model.fit(X_train, y_train)
    adaptive = model.fork() if hasattr(model, "fork") else copy.deepcopy(model)
    predictions = {mode: np.empty(len(y_test)) for mode in ["Static", "Real-Time"]}
    latencies = {mode: [] for mode in predictions}
    update_times = []
    logs = []
    for batch_number, indices in enumerate(BATCHES, start=1):
        xb = X_test[indices]
        yb = y_test[indices]
        # Predict BEFORE using the current batch's target values.
        for mode, current_model in [("Static", model), ("Real-Time", adaptive)]:
            predictions[mode][indices] = current_model.predict(xb)
            latencies[mode].extend(single_sample_latencies(current_model, xb))
        if batch_number == 1:
            assert np.allclose(predictions["Static"][indices], predictions["Real-Time"][indices])
        logs.append({
            "Batch": batch_number,
            "Time_s": test_df.iloc[indices].time_s.max(),
            "Static_RMSE": regression_metrics(yb, predictions["Static"][indices])["RMSE"],
            "RealTime_RMSE": regression_metrics(yb, predictions["Real-Time"][indices])["RMSE"],
        })
        # Skip the final update because no future predictions follow.
        if batch_number < len(BATCHES):
            started = perf_counter()
            adaptive.update(xb, yb)
            update_times.append(perf_counter() - started)
        print(f"  Batch {batch_number}/{len(BATCHES)} completed", flush=True)
    # Verify that the Static model remained unchanged.
    assert np.allclose(model.predict(X_test[:8]), predictions["Static"][:8])
    RESULTS = [row for row in RESULTS if row["Model"] != name]
    for mode in predictions:
        RESULTS.append({
            "Model": name,
            "Mode": mode,
            **regression_metrics(y_test, predictions[mode]),
            "PredictionLatency_ms": np.median(latencies[mode]),
            "AverageUpdateTime_s": np.mean(update_times) if mode == "Real-Time" else np.nan,
            "Updates": len(update_times) if mode == "Real-Time" else 0,
            "UpdateMethod": update_method if mode == "Real-Time" else "Frozen model",
        })
    PREDICTIONS[name] = predictions
    BATCH_LOGS[name] = pd.DataFrame(logs)
    columns = ["Mode", "MAE", "RMSE", "R2", "Bias", "PredictionLatency_ms", "AverageUpdateTime_s"]
    with pd.option_context("display.max_columns", None, "display.float_format", "{:.8g}".format):
        display(pd.DataFrame([row for row in RESULTS if row["Model"] == name])[columns])
    return model
print("Training rows:", len(y_train), "| Test rows:", len(y_test))
print("Sensor batches:", len(BATCHES), "| Updates:", len(BATCHES) - 1)
```

### Model 3 — Convolutional Neural Network Regression:

```python
# MODEL 3 — CONVOLUTIONAL NEURAL NETWORK REGRESSION
class ConvolutionalNeuralNetworkRegressor:
    def build_model(self, learning_rate):
        model = tf.keras.Sequential([
            tf.keras.layers.Input(shape=(len(FEATURES), 1)),
            tf.keras.layers.Conv1D(32, 2, activation="relu"),
            tf.keras.layers.Conv1D(16, 2, activation="relu"),
            tf.keras.layers.Flatten(),
            tf.keras.layers.Dense(32, activation="relu"),
            tf.keras.layers.Dense(1),
        ])
        model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate), loss="mse")
        return model
    def fit(self, X, y):
        mask = INITIAL_FIT_MASK
        tf.keras.utils.set_random_seed(SEED)
        sx = StandardScaler().fit(X[mask])
        sy = StandardScaler().fit(y[mask, None])
        trial = self.build_model(1e-3)
        history = trial.fit(
            sx.transform(X[mask])[..., None],
            sy.transform(y[mask, None]),
            validation_data=(sx.transform(X[~mask])[..., None], sy.transform(y[~mask, None])),
            epochs=80, batch_size=128, shuffle=True, verbose=0,
            callbacks=[tf.keras.callbacks.EarlyStopping(monitor="val_loss", patience=8)],
        )
        self.epochs_ = int(np.argmin(history.history["val_loss"]) + 1)
        # Rebuild and train on the complete initial training set.
        tf.keras.backend.clear_session()
        tf.keras.utils.set_random_seed(SEED)
        self.sx = StandardScaler().fit(X)
        self.sy = StandardScaler().fit(y[:, None])
        self.model = self.build_model(1e-3)
        self.model.fit(
            self.sx.transform(X)[..., None],
            self.sy.transform(y[:, None]),
            epochs=self.epochs_, batch_size=128, shuffle=True, verbose=0,
        )
        self.replay_X = X[-2000:].copy()
        self.replay_y = y[-2000:].copy()
        self.rng = np.random.default_rng(SEED)
        print("Selected training epochs:", self.epochs_)
        return self
    def predict(self, X):
        predicted = self.model(self.sx.transform(X)[..., None], training=False).numpy()
        return self.sy.inverse_transform(predicted).ravel()
    def fork(self):
        other = ConvolutionalNeuralNetworkRegressor()
        other.sx = copy.deepcopy(self.sx)
        other.sy = copy.deepcopy(self.sy)
        other.model = other.build_model(1e-4)
        other.model.set_weights(self.model.get_weights())
        other.replay_X = self.replay_X.copy()
        other.replay_y = self.replay_y.copy()
        other.rng = np.random.default_rng(SEED)
        return other
    def update(self, X, y):
        chosen = self.rng.choice(len(self.replay_y), min(512, len(self.replay_y)), replace=False)
        mixed_X = np.vstack([X, self.replay_X[chosen]])
        mixed_y = np.concatenate([y, self.replay_y[chosen]])
        xt = self.sx.transform(mixed_X)[..., None]
        yt = self.sy.transform(mixed_y[:, None])
        for _ in range(2):
            order = self.rng.permutation(len(yt))
            for start in range(0, len(order), 128):
                indices = order[start:start + 128]
                self.model.train_on_batch(xt[indices], yt[indices])
        self.replay_X = np.vstack([self.replay_X, X])[-2000:]
        self.replay_y = np.concatenate([self.replay_y, y])[-2000:]
convolutional_model = evaluate_model(
    "Convolutional Neural Network Regression",
    ConvolutionalNeuralNetworkRegressor(),
    "Two fine-tuning epochs with replay; fixed scalers",
)
```

### Model 4 — Mixture Gaussian Process Regression:

```python
# MODEL 4 — MIXTURE GAUSSIAN PROCESS REGRESSION
class MixtureGaussianProcessRegressor:
    def __init__(self, window=400):
        self.window = window
    @staticmethod
    def training_optimizer(objective, initial_theta, bounds):
        result = minimize(
            objective, initial_theta, method="L-BFGS-B", jac=True, bounds=bounds,
            options={"maxiter": 150, "maxls": 50, "ftol": 1e-6, "gtol": 1e-5},
        )
        if not result.success:
            warnings.warn(f"Gaussian process optimiser: {result.message}")
        return result.x, result.fun
    def fit(self, X, y):
        self.scaler = StandardScaler().fit(X)
        z = self.scaler.transform(X)
        self.gate = GaussianMixture(n_components=2, n_init=3, random_state=SEED).fit(z)
        groups = self.gate.predict(z)
        self.experts = {}
        self.buffers = {}
        self.kernels = {}
        for k in range(2):
            indices = np.flatnonzero(groups == k)
            if len(indices) > self.window:
                indices = indices[np.linspace(0, len(indices) - 1, self.window, dtype=int)]
            kernel = ConstantKernel(1.0, (1e-3, 1e3)) * RBF(np.ones(X.shape[1]), (1e-2, 1e3)) + WhiteKernel(1e-4, (1e-8, 1e-1))
            expert = GaussianProcessRegressor(
                kernel=kernel, alpha=1e-8, normalize_y=True,
                optimizer=self.training_optimizer, random_state=SEED,
            ).fit(z[indices], y[indices])
            self.experts[k] = expert
            self.kernels[k] = copy.deepcopy(expert.kernel_)
            self.buffers[k] = (z[indices].copy(), y[indices].copy())
        print("Total expert training rows:", sum(len(yb) for _, yb in self.buffers.values()))
        return self
    def predict(self, X):
        z = self.scaler.transform(X)
        weights = self.gate.predict_proba(z)
        means = np.column_stack([self.experts[k].predict(z) for k in range(2)])
        return np.sum(weights * means, axis=1)
    def update(self, X, y):
        z = self.scaler.transform(X)
        groups = self.gate.predict(z)
        for k in range(2):
            selected = groups == k
            if not selected.any():
                continue
            bx, by = self.buffers[k]
            bx = np.vstack([bx, z[selected]])[-self.window:]
            by = np.concatenate([by, y[selected]])[-self.window:]
            self.buffers[k] = (bx, by)
            self.experts[k] = GaussianProcessRegressor(
                kernel=copy.deepcopy(self.kernels[k]), optimizer=None,
                alpha=1e-8, normalize_y=True, random_state=SEED,
            ).fit(bx, by)
mixture_gaussian_process_model = evaluate_model(
    "Mixture Gaussian Process Regression",
    MixtureGaussianProcessRegressor(window=400),
    "Refit up to 400 samples per expert; fixed gate, scaler and kernels",
)
```

### Model 6 — Multilayer Perceptron Regression:

```python
# MODEL 6 — MULTILAYER PERCEPTRON REGRESSION
class MultilayerPerceptronRegressor:
    def build_model(self):
        return MLPRegressor(
            hidden_layer_sizes=(64, 32), activation="relu", solver="adam",
            alpha=1e-4, learning_rate_init=1e-3, batch_size=128,
            early_stopping=False, shuffle=True, random_state=SEED,
        )
    def fit(self, X, y):
        mask = INITIAL_FIT_MASK
        sx = StandardScaler().fit(X[mask])
        sy = StandardScaler().fit(y[mask, None])
        xt = sx.transform(X[mask])
        yt = sy.transform(y[mask, None]).ravel()
        xv = sx.transform(X[~mask])
        yv = sy.transform(y[~mask, None]).ravel()
        trial = self.build_model()
        best_loss = np.inf
        patience = 0
        self.epochs_ = 1
        for epoch in range(1, 201):
            trial.partial_fit(xt, yt)
            validation_loss = mean_squared_error(yv, trial.predict(xv))
            if validation_loss < best_loss - 1e-6:
                best_loss = validation_loss
                self.epochs_ = epoch
                patience = 0
            else:
                patience += 1
            if patience >= 15:
                break
        # Rebuild and train on the complete initial training set.
        self.sx = StandardScaler().fit(X)
        self.sy = StandardScaler().fit(y[:, None])
        xt = self.sx.transform(X)
        yt = self.sy.transform(y[:, None]).ravel()
        self.model = self.build_model()
        for _ in range(self.epochs_):
            self.model.partial_fit(xt, yt)
        self.replay_X = X[-2000:].copy()
        self.replay_y = y[-2000:].copy()
        self.rng = np.random.default_rng(SEED)
        print("Selected training epochs:", self.epochs_)
        return self
    def predict(self, X):
        predicted = self.model.predict(self.sx.transform(X))
        return self.sy.inverse_transform(predicted[:, None]).ravel()
    def update(self, X, y):
        chosen = self.rng.choice(len(self.replay_y), min(512, len(self.replay_y)), replace=False)
        mixed_X = np.vstack([X, self.replay_X[chosen]])
        mixed_y = np.concatenate([y, self.replay_y[chosen]])
        xt = self.sx.transform(mixed_X)
        yt = self.sy.transform(mixed_y[:, None]).ravel()
        for _ in range(2):
            self.model.partial_fit(xt, yt)
        self.replay_X = np.vstack([self.replay_X, X])[-2000:]
        self.replay_y = np.concatenate([self.replay_y, y])[-2000:]
multilayer_perceptron_model = evaluate_model(
    "Multilayer Perceptron Regression (Artificial Neural Network)",
    MultilayerPerceptronRegressor(),
    "Two partial_fit epochs with replay; fixed scalers",
)
```
